```python


import random
import math
def get_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def print_board(state):
    for row in range(8):
        print(" ".join("Q" if state[col] == row else "." for col in range(8)))



def simulated_annealing(initial_temp=100.0, cooling_rate=0.995, min_temp=0.001):

    current_state = [6, 2, 7, 1, 4, 0, 3, 5]
    current_cost = get_conflicts(current_state)

    T = initial_temp

    history = [(0, current_cost, T)]
    iteration = 0

    while T > min_temp and current_cost > 0:
        iteration += 1

        neighbor = get_random_neighbor(current_state)
        neighbor_cost = get_conflicts(neighbor)

        delta_E = current_cost - neighbor_cost

        if delta_E > 0:
            current_state = neighbor
            current_cost = neighbor_cost
        else:
            probability = math.exp(delta_E / T)
            if random.random() < probability:
                current_state = neighbor
                current_cost = neighbor_cost

        T *= cooling_rate
        history.append((iteration, current_cost, T))

    return current_state, current_cost, history


def get_random_neighbor(state):
    neighbor = list(state)
    col = random.randint(0, 7)
    current_row = state[col]

    new_row = random.randint(0, 7)
    while new_row == current_row:
        new_row = random.randint(0, 7)

    neighbor[col] = new_row
    return neighbor


if __name__ == "__main__":
    final_state, final_cost, run_history = simulated_annealing()

    print(f"Final State Array: {final_state}")
    print(f"Remaining Conflicts: {final_cost}")
    print(f"Total Iterations: {run_history[-1][0]}\n")

    if final_cost == 0:
        print("Global Optimum Found")
        print_board(final_state)
    else:
        print("Terminated at a local minimum. Run again to try a new path.")
```
