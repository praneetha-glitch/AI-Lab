```python
def is_var(x):
    return x in ['x', 'y', 'z', 'X', 'Y', 'Z']


def unify(a, b, subs):
    if a == b:
        return True

    if is_var(a):
        subs[a] = b
        return True

    if is_var(b):
        subs[b] = a
        return True

    if '(' in a and '(' in b:
        f1 = a[:a.index('(')]
        f2 = b[:b.index('(')]

        if f1 != f2:
            return False

        a1 = a[a.index('(')+1:-1].split(',')
        b1 = b[b.index('(')+1:-1].split(',')

        if len(a1) != len(b1):
            return False

        for x, y in zip(a1, b1):
            if not unify(x.strip(), y.strip(), subs):
                return False

        return True

    return False


a = input("Enter first expression: ")
b = input("Enter second expression: ")

subs = {}

if unify(a, b, subs):
    print("Unification Successful")
    print("Substitution:", subs)
else:
    print("Unification Failed")
```
