```python
rooms = {
    'A': input("Enter status of Room A (Dirty/Clean): ").capitalize(),
    'B': input("Enter status of Room B (Dirty/Clean): ").capitalize()
}
location = input("Enter initial location of vacuum (A/B): ").upper()

def simple_reflex_agent(location, status):

    if status == "Dirty":
        return "Suck"

    elif location == "A":
        return "Right"

    else:
        return "Left"

print("\nInitial State:")
print("Room A:", rooms['A'])
print("Room B:", rooms['B'])
print("Vacuum Location:", location)

steps = 0

while rooms['A'] == "Dirty" or rooms['B'] == "Dirty":

    status = rooms[location]

    action = simple_reflex_agent(location, status)

    print("\nCurrent Room:", location)
    print("Current Status:", status)
    print("Action:", action)

    if action == "Suck":
        rooms[location] = "Clean"

    elif action == "Right":
        location = "B"

    elif action == "Left":
        location = "A"

    steps += 1

print("Final State:")
print("Room A:", rooms['A'])
print("Room B:", rooms['B'])
print("Vacuum Location:", location)

print("Number of actions:", steps)
```
