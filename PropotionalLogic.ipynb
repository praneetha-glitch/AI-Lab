```python



import re

def PL_TRUE(expr, model):
    expr = expr.replace('&', ' and ')
    expr = expr.replace('|', ' or ')
    expr = expr.replace('~', ' not ')
    expr = expr.replace('->', ' IMPLIES ')
    expr = expr.replace('<->', ' IFF ')

    expr = re.sub(r'([A-Z])\s*IMPLIES\s*([A-Z])',
                  r'((not \1) or \2)', expr)
    expr = re.sub(r'([A-Z])\s*IFF\s*([A-Z])',
                  r'(\1 == \2)', expr)

    for s, v in model.items():
        expr = re.sub(r'\b' + s + r'\b', str(v), expr)

    return eval(expr)


def TT_CHECK_ALL(KB, alpha, symbols, model):

    if not symbols:
        if PL_TRUE(KB, model):
            return PL_TRUE(alpha, model)
        return True

    P = symbols[0]
    rest = symbols[1:]

    return (TT_CHECK_ALL(KB, alpha, rest, {**model, P: True})
            and
            TT_CHECK_ALL(KB, alpha, rest, {**model, P: False}))


def TT_ENTAILS(KB, alpha, symbols):
    return TT_CHECK_ALL(KB, alpha, symbols, {})


KB = input("Enter KB: ")
alpha = input("Enter Query: ")

symbols = sorted(set(re.findall(r'\b[A-Z]\b', KB + alpha)))

if TT_ENTAILS(KB, alpha, symbols):
    print("KB entails alpha")
else:
    print("KB does not entail alpha")




```
