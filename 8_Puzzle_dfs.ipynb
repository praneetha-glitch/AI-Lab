```python
goal = "123456780"

def print_state(state):
    print(state[0:3])
    print(state[3:6])
    print(state[6:9])
    print()

def dfs(start):

    stack = [start]
    visited = set()

    while stack:

        state = stack.pop()

        if state in visited:
            continue

        visited.add(state)

        print("State", len(visited))
        print_state(state)

        if state == goal:
            print("Goal Found!")
            return

        zero = state.index("0")

        row = zero // 3
        col = zero % 3
        moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

        for r, c in moves:

            new_row = row + r
            new_col = col + c

            if 0 <= new_row < 3 and 0 <= new_col < 3:

                new_zero = new_row * 3 + new_col

                new_state = list(state)

                new_state[zero], new_state[new_zero] = \
                    new_state[new_zero], new_state[zero]

                new_state = "".join(new_state)

                if new_state not in visited:
                    stack.append(new_state)


start = input("Enter puzzle: ")

dfs(start)
```
